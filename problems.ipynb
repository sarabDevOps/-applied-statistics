 #### The notebook looks for iris.csv beside itself first, change DATA_PATH to the location of the CSV file 

```python
import pandas as pd
import numpy as np      
import matplotlib.pyplot as plt
from pathlib import Path    



DATA_PATH = Path('iris.csv')
if not DATA_PATH.exists():
    DATA_PATH = Path(r'C:\Users\sarab\Downloads\iris.csv')

iris = pd.read_csv(DATA_PATH)
required = {'petal_length', 'species'}
missing = required.difference(iris.columns)
if missing:
    raise ValueError(f'Missing required columns: {missing}')    


iris['species'] = iris['species'].str.strip().str.lower()
print(f'Rows: {len(iris)}')
print('Species counts:')
display(iris['species'].value_counts().reindex(['setosa', 'versicolor', 'virginica']))
iris.head()

 
```
